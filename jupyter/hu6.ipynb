```python
df_der = pd.read_excel("../../data/hu3_esquizofrenia.xlsx")
df_izquierda = pd.read_csv("../../data/hu5_completo.csv")

df = df_izquierda.merge(
    right=df_der[['nregistro','Precio de referencia', 'num_principios_activos']],
    right_on="nregistro",
    left_on="nregistro",
)

df.head()
df.isna().sum()
```

```python
mediana_pa_forma = df.groupby(["pa_base", "forma_farmaceutica_simplificada"])["Precio de referencia"].transform("median")
mediana_forma = df.groupby("forma_farmaceutica_simplificada")["Precio de referencia"].transform("median")

df["Precio de referencia"] = df["Precio de referencia"].fillna(mediana_pa_forma)
df["Precio de referencia"] = df["Precio de referencia"].fillna(mediana_forma)
df["Precio de referencia"] = df["Precio de referencia"].fillna(df["Precio de referencia"].median())

df.isna().sum()
```

```python
df.to_csv("../../data/hu5_completo_v2.csv", index=False)
```
